# Watermark catalog — getting started

Thermographic recordings of watermarks in music manuscripts and prints, published on [Zenodo](https://zenodo.org/). `wmcat` gives you the catalog as a pandas DataFrame and loads individual recordings on demand.

This notebook runs as-is in the browser (JupyterLite) and locally.

````python
from wmcat import open_catalog

cat = open_catalog()
cat
````

````python
# Substring match (case-insensitive) on any column
lilies = cat.search(motif="lily")
print(lilies)

# Exact match: '01f' would otherwise also match combinations like '01f+02f'
print(cat.search(wm_type="01f", exact=True))
````

````python
# Load one recording (~4.6 MB) from Zenodo and display it
sub = cat.search(holding_institution="SBB", motif="Krone")
print(sub)
````

````python
w = await sub.aload(0)
w.show();
````

````python
w.meta["doi"], w.meta["shelfmark"], w.frames.shape
````
